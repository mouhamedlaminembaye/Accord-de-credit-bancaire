## Introduction Artificial Neural Network

### Import libraries

```python
import pandas as pd
import numpy as np
import tensorflow as tf
```

```python
tf.__version__
```

## Part 1 : Data Preprocessing

#### Import the dataset

```python
dataset = pd.read_excel('/content/sample_data/Folds5x2_pp.xlsx')
dataset.head()
```

```python
X = dataset.iloc[:, :-1].values
y =dataset.iloc[:, -1].values
```

```python
X, y
```

#### Split the dataset into training and test set

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

## Part 2 : Building the ANN

#### Initialize the ANN

```python
# Initialize the ann as  sequence of layers
ann = tf.keras.models.Sequential()
```

#### Adding the input layer and the hidden layer

```python
ann.add(tf.keras.layers.Dense(units=6, activation='relu'))
```

#### Adding the second hidden layers

```python
ann.add(tf.keras.layers.Dense(units=6, activation='relu'))
```

#### Adding the output layer

```python
ann.add(tf.keras.layers.Dense(units=1))
```

## PArt 3 : Train the ANN

#### Compile the AN

```python
ann.compile(optimizer='adam', loss='mean_squared_error')
```

#### Train the model

```python
ann.fit(X_train, y_train, epochs=100, batch_size=32, validation_data=(X_test, y_test))
```

#### Predicting the result of the tesrt set

```python
y_pred = ann.predict(X_test)
np.set_printoptions(precision=2)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```
