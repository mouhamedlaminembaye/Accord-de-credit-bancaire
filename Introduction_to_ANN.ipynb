# Introduction to ANN : Building a classifcation model

### Import libraries

```python
import numpy as np
import pandas as pd
import tensorflow as tf
```

```python
tf.__version__
```

# Part 1 : Data preprocessing

```python
data = pd.read_csv('/content/sample_data/Churn_Modelling.csv')
data.head()
```

```python
data.shape
```

```python
X = data.iloc[:, 3:-1].values
y = data.iloc[:, -1].values
```

```python
print(X)
```

```python
print(y)
```

### Ecoding categorical variable

##### Label encoder : the 'Gender' column

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
X[:, 2] = le.fit_transform(X[:, 2])
```

```python
X
```

##### One hot encoding : the Geography column

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [1])], remainder='passthrough')
X = np.array(ct.fit_transform(X))
```

```python
X
```

### Split the dataset into train set and test set

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

### Feature scaling

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
X
```

# Part 2 : Building the ANN

##### Initialize the ann

```python
ann = tf.keras.models.Sequential()
```

##### Add the input layer and the first hidden layer

```python
ann.add(tf.keras.layers.Dense(units=6, activation='relu'))
```

##### Add the second hidden layer

```python
ann.add(tf.keras.layers.Dense(units=6, activation='relu'))
```

##### Add the output layer

```python
ann.add(tf.keras.layers.Dense(units=1, activation='sigmoid'))
```

# Part 3 : Train the ANN

##### Compile the ann

```python
ann.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

##### Train the ann on the trainig set

```python
ann.fit(X_train, y_train, batch_size=32, epochs=100)
```

# Part 4 : Predict the result of a single observation

```python
print(ann.predict(sc.transform([[1, 0, 0, 600, 1, 40, 3, 60000, 2, 1, 1, 50000]])) > 0.5)
```

It is clear that for this given customer, the model predicts that probalbly he wont leave the bank

##### Predict the test set result

```python
y_pred = ann.predict(X_test)
y_pred = (y_pred > 0.5)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

##### Make the confusion matrix

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
print(cm)
accuracy_score(y_test, y_pred)
```
